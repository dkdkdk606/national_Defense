## 2. 지도 조절

- Map()의 zoom_start 인수 : Map을 처음 그릴 때 zoom의 범위 값 설정
- Map()의 control_scale 인수 : 스케일 컨트롤 버튼 표시 여부
- Map()의 zoom_control 인수 : zoom in/out 버튼 표시 여부

#### ① 지도 확대/축소하기 : Map( )의 **zoom_start** 인수

- 화면 확대 비율 조절
- 기본값은 10
- 값이 커질수록 확대 : (+)버튼 기능을 클릭한 것과 같은 효과

````python
folium.Map(location =[lat,lon], zoom_start=3)
````

② control scale 표시 : Map( )의 **control_scale 인수**
- bool, default True

````python
folium.Map(location =[lat,lon], zoom_start=20, zoom_control=False, control_scale=True)
````

③ zoom control 표시 : Map( )의 **zoom_control 인수**
- bool or position string
- default: True (top left corner)
- other options : 'topleft', 'topright', 'bottomleft' or 'bottomright'

````python
folium.Map(location =[lat,lon], zoom_start=20, zoom_control=False, control_scale=True)
````

② control scale 표시 : Map()의 control_scale 인수

거리 축척(scale) 표시 여부 : 미터(m)와 피트(feet) 거리 비율을 보여줌(좌하단 위치)

bool, default True

#### 2) **folium.TileLayer()** 를 이용한 지도 기본 tiles 지정
```python
folium.raster_layers.TileLayer(tiles: str | TileProvider = 'OpenStreetMap', 
            min_zoom: int = 0, max_zoom: int = 18, max_native_zoom: int | None = None, 
            attr: str | None = None, detect_retina: bool = False, 
            name: str | None = None, overlay: bool = False, control: bool = True,
            show: bool = True, no_wrap: bool = False, subdomains: str = 'abc',
            tms: bool = False, opacity: float = 1, **kwargs)
```

https://python-visualization.github.io/folium/latest/reference.html#folium.raster_layers.TileLayer

````python
# 지도스타일 : 'Stamen Terrain'
lat, lon = 37.566535, 126.9779691999996
m = folium.Map(location=[lat, lon], zoom_start=17)

# 배경지도 속성 설정
tiles = 'Stamen Terrain'
attr = 'Map tiles by Stamen Design, under CC BY 3.0. Data by OpenStreetMap, under ODbL.'

# 배경지도 타일에 추가로 타일 설
folium.TileLayer(tiles=tiles, attr=attr).add_to(m)
m
````

#### 3) **custom 타일** 사용하기

- 제공되는 custom tile 미리보기 : http://leaflet-extras.github.io/leaflet-providers/preview

- Map(tiles=custome타일url, attr=provider명)

````python
folium.Map(location=[lat, lon],
           zoom_start=16,
           tiles ='https://tiles.stadiamaps.com/tiles/osm_bright/{z}/{x}/{y}{r}.png',
           attr='Stadia Maps')
````

- **TileLayer(tiles=TilesUrl, attr=str)** 사용하여 custome tiles 지정

````python
m = folium.Map(location=[lat, lon], zoom_start=16)
layer = 'osm_bright'
tile_type = 'png'
tiles_url = f'https://tiles.stadiamaps.com/tiles/{layer}/{{z}}/{{x}}/{{y}}{{r}}.{tile_type}'
attr = 'Stadia Maps'
folium.TileLayer(tiles = tiles_url,
                 attr=attr
                ).add_to(m)
m
````

- **Vworld 오픈API** 이용    

   - 국토교통부 공간정보산업진흥원 브이월드(https://www.vworld.kr/v4po_main.do)
   - 참고. 국토정보플랫폼의 바로e맵(https://map.ngii.go.kr/mi/emapMain/emapIntro01.do)
   - 타일종류 / 타일 타입
      - Base / png
      - gray / png
      - midnight / png
      - Hybrid / png
      - Satellite / jpeg

````python
lat, lon = 37.556047, 126.923019

m = folium.Map(location=[lat, lon], zoom_start=16)
vworld_key = 'FDC4288B-CA56-4FC5-9A4A-4F4FD00F4679'
layer = 'Base'
tile_type = 'png'
tiles_url = f'https://api.vworld.kr/req/wmts/1.0.0/{vworld_key}/tiles/{layer}/{{z}}/{{y}}/{{x}}.{tile_type}'
attr = 'VWorld'

folium.TileLayer(tiles = tiles_url,
                 attr=attr
                ).add_to(m)
m
````

#### 구글 지도 설정

지도 타일명 : lyrs=값str
- Standard Road Map : m
- Terran Map : p
- Somehow Altered Road Map : r
- Satellite Map : s
- Terran Only : t
- Hybrid : y
- Roads Only : h

````python
# Google 지도스타일 => Standard Road Map

lat, lon = 37.556047, 126.923019  # 홍대지하철 9번출구
m = folium.Map(location=[lat, lon], zoom_start=16)

vworld_key = 'CDBFC2A1-D9AB-4926-9816-8E8070C2F4C1'
layer = 'Satellite'
tile_type = 'jpeg'
tiles_url = f'http://api.vworld.kr/req/wmts/1.0.0/{vworld_key}/{layer}/{{z}}/{{y}}/{{x}}.{tile_type}'
attr = 'VWorld'

folium.TileLayer(tiles = tiles_url,
attr=attr,
overlay=True,
control=True,
).add_to(m)
m
````
