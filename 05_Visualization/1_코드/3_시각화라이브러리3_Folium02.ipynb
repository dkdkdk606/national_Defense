## 5. 지도에 마커(marker) 설정 

- POI(Point Of Interest, 관심지점)을 추가하고 관련 정보 제공
- 마커(marker) : 특정 위치를 표시하는 표식

### 마커를 부착한 지도 생성

단계1. 마커 생성   
- **folium.Marker(location=[위도,경도], popup='', tooltip='', icon=아이콘객체)**
   - popup : 마커 클릭 시 표시되는 문구
       - 환경에 따라 지원 안되는 경우도 있음
   - tooltip : 마우스 hover시 표기되는 문구
   - icon : Marker 형태 결정
       - 기본값 : information icon
                
단계2. 마커 아이콘 생성
- **folium.Icon(color='색상', icon='아이콘 모양')**

    
   
단계3. 마커 생성 후 부착
- **folium.Marker().add_to(지도객체변수)**

## 6. 지도에 원 그리기

: 지도에 데이터 값, 영역 등의 정보를 표현하기 위한 방법

- folium.Circle()
```python
folium.Circle(
    location: Sequence[float] | None = None,
    radius: float = 50,
    popup: folium.map.Popup | str | None = None,
    tooltip: folium.map.Tooltip | str | None = None,
    **kwargs: bool | str | float | None,
)
```
  
- folium.CircleMarker()
```python
folium.CircleMarker(
    location: Sequence[float] | None = None,
    radius: float = 10,
    popup: folium.map.Popup | str | None = None,
    tooltip: folium.map.Tooltip | str | None = None,
    **kwargs: bool | str | float | None,
)
```

#### Circle()  생성

````python
zoom=15
m=folium.Map(location=hongdae, zoom_start=zoom)

#Circle
folium.Circle(location = hongdae,
              radius=100, # 원 반지름 크기
              color='red', # 원 선 색상
              fill_color='blue', # 원 내부 색상
              popup='Circle popup',
              tooltip='Circle tooltip'
              
             ).add_to(m)
````

````python
zoom=15
m=folium.Map(location=hongdae, zoom_start=zoom, control_scale=True)

#Circle
folium.Circle(location = hongdae,
              radius=100, # 원 반지름 크기
              color='red', # 원 선 색상
              fill_color='blue', # 원 내부 색상
              popup='Circle popup',
              tooltip='Circle tooltip'
              
             ).add_to(m)
````

````python
m
````

#### CircleMarker 생성

````python
zoom=15
m=folium.Map(location=hongdae, zoom_start=zoom, control_scale=True)

#Circle
folium.CircleMarker(location = hongdae,
                      radius=50, # 원 반지름 크기
                      color='#34ebc6', # 원 선 색상 코드도 되고 정해져있는 색상도 가능
                      fill_color='blue', # 원 내부 색상  //
                      popup='CircleMarker popup',
                      tooltip='CircleMarker tooltip'
                      
                     ).add_to(m)
````

````python
m
````

Circle() 과 CircleMarker() 의 차이점
- Circle()
    - 단위 : meter
    - 크기변화 : 지도를 확대/축소하면 실제 거리(미터)를 기준으로 화면상에 보이는 원의 크기가 변동됨
    - 용도 : 특정 지점의 실제 영향 반경을 정확하게 면적으로 표시할 때 적합


- CircleMarker()
    - 단위 : 픽셀
    - 크기변화 : 지도를 확대/축소해도 화면상에 보이는 원의 크기가 일정하게 유지
    - 용도: 실제거리가 아니라 단순히 지도 위에 특정 위치를 나타내는 마커 기호 대신 원으로 표시할 때 적합
