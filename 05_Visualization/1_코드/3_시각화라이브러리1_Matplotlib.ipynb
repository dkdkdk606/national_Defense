: *args : x, y 축값을 리스트 형식으로 입력, y축값만 쓰는 경우 x축값은 자동 생성

````python
x = np.linspace(0, np.pi*2, 50)
print(x)
y = np.sin(x)
plt.figure(figsize=(3,3))
plt.plot(x,y)
plt.show()
````

### 2) plot() 에서 사용되는 선 모양 관련  속성(약자로도 표기 가능)

* 선 색깔      : color (c)
* 선 굵기      : linewidth (lw)
* 선 스타일    : linestyle (ls)
* 마커 종류    : marker
* 마커 크기    : markersize (ms)
* 마커 선색깔  : markeredgecolor (mec)
* 마커 선굵기  : markeredgewidth (mew)
* 마커 내부색깔: markerfacecolor (mfc)   

#### ① 선 색 지정 : plt.plot(*args, color='색상명')

````python
x = np.linspace(0, np.pi*2, 50)
y = np.sin(x)
plt.figure(figsize=(3,3))
plt.plot(x,y, lw=4) # linewidth=1
plt.show()
````

#### ② 선 굵기 : plt.plot(*args, linewidth=정수값)

- linewidth 대신 lw 사용

#### ③ 선스타일(line style) 지정 : plt.plot(*args, linestyle=스타일이름)

- linestyle = 'solid' | 'dotted' | 'dashed' | 'dashdot'

- 선스타일 기호
  


https://matplotlib.org/stable/api/_as_gen/matplotlib.lines.Line2D.html#matplotlib.lines.Line2D.set_linestyle

````python
x = np.linspace(0, np.pi*2, 50)
y = np.sin(x)
plt.figure(figsize=(5,3))
for i, ls in enumerate(['solid','dashed','dashdot','dotted']):
    plt.plot(x+i, y, lw=i+1, ls=ls, label=ls)  # linewidth=1, # linestyle
plt.legend()
plt.show()
````

#### ④ 마커 지정 : plt.plot(*args, marker=마커유형)

- marker = 



````python
markers = {'circle':'o', 'plus':'+', 'star':'*',
           'dot':'.', 'cross':'x', 'square':'s',
           'diamond':'d', 'upper tri':'^', 'lower tri':'v',
           'right tri': '>', 'left tri':'<', 'pentagon':'p', 'hexagram':'h'}

x = np.linspace(0, np.pi*2, 20)
y = np.sin(x)
plt.figure(figsize=(12,5))
for i, item in enumerate(markers.items()):
    label = item[0]
    m = item[1]
    plt.plot(x+i, y, marker=m, label=label)  # linewidth=1, # linestyle
plt.legend()
plt.show()
````

#### ⑤ 마커 크기 변경 : plt.plot(*args, makersize=)

- 약어 : ms 사용

````python
markers = {'circle':'o', 'plus':'+', 'star':'*',
           'dot':'.', 'cross':'x', 'square':'s',
           'diamond':'d', 'upper tri':'^', 'lower tri':'v',
           'right tri': '>', 'left tri':'<', 'pentagon':'p', 'hexagram':'h'}

x = np.linspace(0, np.pi*2, 20)
y = np.sin(x)
plt.figure(figsize=(12,5))
for i, item in enumerate(markers.items()):
    label = item[0]
    m = item[1]
    plt.plot(x+i, y, marker=m, label=label, ms= 15-(i+1))  # linewidth=1, # linestyle
plt.legend()
plt.show()
````

#### ⑥ 마커 선굵기, 색상 변경 

- 마커 선색깔 : markeredgecolor (mec)
- 마커 선굵기 : markeredgewidth (mew)
- 마커 내부색깔: markerfacecolor (mfc)

````python
y = [1,4,5,8,9,5,3]
plt.figure(figsize=(5,3))
plt.plot(y, ls='-.', marker='o', ms=8, mfc='blue', mec='r', mew=1)   # markeredgecolor : mec
plt.xlim(-2,8)
plt.ylim(0,10)
plt.title('마커 스타일적용 사례')
plt.show()
````

#### ⑦ 선의 색상과 스타일을 한번에 지정하기
- 선모양 선색상에 해당하는 기호나 문자열을 축약하여 한 줄에 씀
- 예. 'r-.', 'b--'

````python
y = [1,4,5,8,9,5,3]
x = np.arange(0, len(y))
plt.figure(figsize=(5,3))
plt.plot(x, y, 'r-.')  # r(red), -. : dashdot
plt.plot(x+1, y, 'g:')  # g(greed), : dotted
plt.plot(x+2, y, '--y')  # y(yellow), -- : dashed
plt.plot(x+3, y, '-')  # 기본색상, - : solid
plt.xlim(-2,14)
plt.ylim(0,10)
plt.title('선색상과 스타일을 한번에')
plt.show()
````

#### ⑧ 선의 색상과 스타일, 마커모양을 한번에 지정하기

````python
y = [1,4,5,8,9,5,3]
x = np.arange(0, len(y))
plt.figure(figsize=(5,3))
plt.plot(x, y, 'r-.o')  # r(red), -. : dashdot, o :circle
plt.plot(x+1, y, 'g:^')  # g(greed), : dotted, ^ : upper tri
plt.plot(x+2, y, '--yp')  # y(yellow), -- : dashed, p: pentagon
plt.plot(x+3, y, '-*')  # 기본색상, - : solid, *:star
plt.xlim(-2,14)
plt.ylim(0,10)
plt.title('선색상과 스타일, 마커모양을 한번에')
plt.show()
````

### 2) plt.subplots(행,열) 함수

여러 개의 Axes 객체를 동시에 생성해주는 함수

- 행렬 형태의 객체로 반환
- 두 개의 반환값이 있음
    - 첫번 째 반환은 그래프 객체 전체 이름 : 거의 사용하지 않음
    - 두번 째 반환값에 Axes 객체를 반환함
    - 두번 째 반환값이 필요하므로 반환 값 모두를 반환받아 두번 째 값을 사용해야 함

#### subplot(), subplots(), subplot_mosaic() 비교

||subplot()|subplots()|subplot_mosaic()|
|---|---|---|---|
|코드 방식|subplot을 하나씩 수동 지정|subplot을 배열로 일괄 생성|문자열/딕셔너리 기반으로 유연하게 레이아웃 구성|
|접근 방식|번호로 직접 접근(plt.subplot(1,2,1))|배열의 인덱싱(axes[0,1], axes[1], axes[0][1])|이름(label)으로 접근(ax['A'],ax['B'])|
|반환 값| 단일 Axes Subplot 객체|(Figure, Axes) 튜플(Axes 배열 또는 객체 배열)|(Figure, dict[str, Axes]) 형태로 반환|
|코드 구조|반복문 없이 단순한 코드지만 중복 많음|반복문 활용 가능. 배열 인덱싱 필요|레이아웃 정의가 직관적이고 가독성 좋음. 반복문도 가능|
|레이아웃 유연성|낮음|중간(격자형만 가능)|매우 높음(불균형, 중첩, 특정 영역 확대 등 자유롭게 설정 가능)|
|권장 사용|아주 단순한 1xN 또는 N행의 그래프|다수 subplot을 일정한 행렬 형태로 반복하여 그릴 때|복잡한 레이아웃, 대시보드 구성, 이름 기반으로 명확히 구분할 때|
|버전 요구사항|오래된 버전부터 사용 가능|오래된 버전부터 사용 가능|Matplotlib 3.4 이상|
|예시 코드|plt.subplot(2,1,1)|fig, axes = plt.subplots(2,2)|fig, axes = plt.subplot_mosaic([['A','B'],['C','B']]))|
|요약|subplot 하나씩 단순하게 붙이기|반복적인 플롯 그리기|이름으로 구역 구분, 복잡한 배치|
