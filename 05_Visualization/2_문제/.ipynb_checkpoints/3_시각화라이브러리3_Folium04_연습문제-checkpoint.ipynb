### 연습문제1. 서울시 종로구 옥인동에 있는 공중화장실 위치를 찾아 Marker 표시하기

- 데이터 : "서울시 공중화장실 위치정보.csv"
- 출처 : 서울시 열린데이터 광장
- https://data.seoul.go.kr/dataList/OA-22586/S/1/datasetView.do

```python
data = pd.read_csv('data/서울시 공중화장실 위치정보.csv', encoding='cp949')
data.head()
```

```python
data.info()
```

```python
toilet_ok = data[ data['지번주소'].str.contains('옥인동') ]

ok_loc = (37.5835,126.9634)
toilet_ok.head(1)
```

```python
m = folium.Map(location=ok_loc )

for idx, tiolet in toilet_ok.iterrows():
    folium.Marker(location=(tiolet['y 좌표'],tiolet['x 좌표']),
                  popup=tiolet['건물명'],
                  tooltip=tiolet['지번주소'],
                  icon=folium.Icon(color='gray'),
                 ).add_to(m)
    
m
```

### 연습문제2. 서울시 지진대피소 위치를 MarkerCluster로 표시하기

```python
from folium.plugins import MarkerCluster
```

- 데이터 : '서울시 지진옥외대피소.csv'
- 대피소별 주소, 면적, 표값 제공(좌표계: 위경도(GPS), XY좌표(GRS90-중부원점))
- 출처 : 서울시 열린데이터광장
-  https://data.seoul.go.kr/dataList/OA-21063/S/1/datasetView.do

```python
data = pd.read_csv('data/서울시 지진옥외대피소.csv', encoding='cp949')
data.head()
```

```python
data.info()
```

```python
data.head(1)
```

```python
seoul_loc = (37.5666,126.9784) # 서울시

color = 'darkred'
icon = 'info-sign'
prefix='fa'

m = folium.Map(location=seoul_loc)
```

```python

mc = MarkerCluster().add_to(m)
for i, d in data.iterrows():
    folium.Marker(location=(d['위도'],d['경도']),
                  tooltip=d['수용시설명'],
                  popup=d['상세주소'],
                  icon=folium.Icon(color=color,
                                   icon=icon,
                                   prefix=prefix,
                                  ),
                 ).add_to(mc)


m
```

### 연습문제3. 서울시 따릉이 대여소 위치를 MarkerCluster로 표시하기

- 서울시 공공자전거 따릉이 대여소 정보
- 데이터 : '공공자전거 대여소 정보(26.6월 기준).xlsx'
- 출처 : 서울시 열린데이터광장
- https://data.seoul.go.kr/dataList/OA-13252/F/1/datasetView.do
- 주의. 데이터 로딩시 데이터가 있는 시트명 지정하여 열기

```python
data = pd.read_excel('data/공공자전거 대여소 정보(26.6월 기준).xlsx',
                     sheet_name='대여소현황',
                     header=4,
                    )
data.columns = ['대여소ID','보관소(대여소)명','자치구','상세주소','위도','경도','설치시기','설치형태_LCD_거치대수','설치형태_QR_거치대수','운영방식']
data.head(5)
```

```python
seoul_loc = (37.5666,126.9784) # 서울시

color = 'blue'
icon = 'bicycle'
prefix='fa'

m = folium.Map(location=seoul_loc)
```

```python

mc = MarkerCluster().add_to(m)
for i, d in data.iterrows():
    folium.Marker(location=(d['위도'],d['경도']),
                  tooltip=d['보관소(대여소)명'],
                  popup=d['상세주소'],
                  icon=folium.Icon(color=color,
                                   icon=icon,
                                   prefix=prefix,
                                  ),
                 ).add_to(mc)


m
```

### 연습문제4. 대한민국 시도별 1인당 소득 현황을 지도로 시각화하기

- 지도 경계 데이터 : '시도별geo.json'
- 표시 데이터 : '시도별_1인당소득.csv'

```python
file_name = 'data/시도별geo.json'
with open(file_name, encoding='utf-8') as f:
    earn_kr = json.loads(f.read())
```

```python
earn_kr.keys()
```

```python
earn_kr['features'][:2]

  # 'properties': {'CTPRVN_CD': '42',
#    'CTP_ENG_NM': 'Gangwon-do',
#    'CTP_KOR_NM': '강원도'}},
```

```python
earn_df = pd.read_csv('data/시도별_1인당소득.csv')
earn_df.tail(3)
```

```python
earn_df['1인당 개인소득'] = earn_df['1인당 개인소득'].str.replace(',','').astype(float)
```

```python
earn_df.info()
```

```python
# 시도별 1인당 소득수준
loc_kr = (37, 127)  # 한국 중심 쯤
zoom = 8
m = folium.Map(location=loc_kr, zoom_start=zoom)

folium.GeoJson(earn_kr).add_to(m)

folium.Choropleth(
    geo_data=earn_kr,   #지도경계 구분을 위한 json
    data=earn_df,    #지도에 표시할 데이터프레임
    columns=['시도별', '1인당 개인소득'], #지도에 표시할 data의 컬럼
    key_on='feature.properties.CTP_KOR_NM',   # geo json의 지역코드 : feature속성의 지역코드 속성 이름
    fill_color='YlOrRd',
    fill_opacity=0.7,
    line_opacity=1,
    name='choropleth',
    legend_name='1인당 개인 소득',
    
).add_to(m)

folium.LayerControl().add_to(m)

m
```
