### 데이터프레임의 기본 인덱싱

1. 열인덱싱 
    - 하나의 열만 선택 : **` df['열이름'] `**
    - 여러 개 열 선택  : **` df[['열이름1', '열이름2',...]] `**


2. 행인덱싱 : 연속된 구간의 행데이터 선택(슬라이싱)
    - **`df['행_시작위치':'행_끝위치']`** 


3. 개별요소 접근 : 선택한 열에서 지정된 구간의 행데이터 선택
    - **`df['열이름']['행_시작위치':'행_끝위치']`**
    - **`df['열이름']['시작_행이름':'끝_행이름']`**

### 데이터프레임의 특별한 인덱서(indexer) : loc, iloc

- numpy와 같이 쉼표(,)를 사용하여 행과 열을 동시에 인덱싱하는 2차원 인덱싱
- [행 인덱스, 열 인덱스] 형식

- **loc** : 라벨값 기반의 2차원 인덱싱(명칭기반 인덱싱)
- **iloc** : 순서를 나타내는 정수 기반의 2차원 인덱싱(위치기반 인덱싱)

## 1. loc 인덱서

**명칭기반 인덱서**

1) **df.loc[행인덱싱 값]**   # 행우선 인덱서

2) **df.loc[행인덱싱 값,열인덱싱 값]**

**인덱싱 값 유형**

- 인덱스 데이터 : [index name] 또는 [index_name, column name]
- 인덱스 데이터 슬라이스 
- 같은 행 인덱스를 갖는 불리언 시리즈(행 인덱싱인 경우)
     - 조건으로 추출 가능(불린 인덱싱)
- 위 값을 반환하는 함수    

```python
df = pd.DataFrame(data = np.arange(10,22).reshape(3,4),
                  index= list('abc'),
                  columns = list('ABCD')
                 )
df                 
```

```python
# 기본인덱싱 : 열우선
df['A'] # 열 인덱싱 시리즈 반환
df.C
df[['A','C']] # 데이터 프레임 반환
df['a':'b']   # 행 인덱싱 -> 데이터 프레임 반환
df['B']['c'] # 개별 요소 반환
```

### 1) 데이터프레임의 행 선택

**형식 : dataframe.loc['행인덱스값']**

```python
# 'a'행 선택
df['a':'a'] #데이터 프레임 반환
df.loc['a'] # 시리즈 반환
```

#### 주의. loc 인덱서에서는 열 단독 인덱싱은 불가능

### 2) 데이터프레임의 여러 행 선택

#### ① 행인덱스 슬라이싱

**형식: dataFrame.loc['처음 행인덱스값' : '끝행인덱스값']**

- b행부터 c행의 모든 열 반환

```python
df['b':'c']
```

```python
df.loc['b':'c']
```

```python
df.loc[['b','c']]
```

#### 행인덱스가 정수위치인덱스인 경우

```python
df2 = pd.DataFrame(data = np.arange(10,26).reshape(4,4),
                   columns = list('abcd')
                  )
```

```python
df2.loc[2]
```

값인덱싱 - 슬라이싱 초기값:끝값

```python
df2.loc[2:3]              #   -> '2'라는 인덱스 행 부터 '3' 인덱스 행 까지
```

나중에 iloc 쓸 때 -> range(2,3) 에 포함된 행

```python
df2.iloc[2:3]
```

#### ② 행인덱스 데이터를 리스트로 지정

**형식: dataFrame[[행이름1, 행이름2,... ]]**

- 여러 행 선택시 인덱서 데이터를 리스트로 사용
- 반환값이 데이터프레임이 됨

```python
df2.loc[[0,2]]
```

※ 참고. 데이터프레임 기본 인덱싱은 **열기준**

#### ③ boolean indexing으로 행 선택

**조건 식 수행**

```python
df
```

```python
df['A']>15
df[df['A']>15]
```

```python
# loc인덱서 사용 # 열 기준으로 계산할 경우에는 loc 안쓰는게 더 짧긴 
df.loc[:,'A'] >15
df[df.loc[:,'A'] >15]
```

#### ④ 인덱스 대신 인덱스 값을 반환하는 함수 사용

```python
def sel_row(df):
    return df.A > 15
```

```python
df.loc[sel_row(df)]
```

```python
df[ lambda df: df.A>15 ]
```

## 2. iloc 인덱서

- **위치기반 인덱스**
- 라벨(name)이 아닌 위치를 나타내는 정수 인덱스만 사용
- 위치 정수값은 0부터 시작
- **형식 : 데이터프레임.iloc[행, 열]**

- **데이터프레임의 기본 인덱싱은 [행번호,열번호] 인덱싱 불가**

- **iloc[] 사용하면 가능**
    - iloc[행번호,열번호]
    - loc[행이름,열이름]

### 1) iloc을 이용하여 개별요소 선택

형식 : 데이터프레임.iloc[행번호,열번호]

결과값 : **스칼라값 반환됨**

- 0행 1열 선택

```python
df['B']['a']
df['B']['a':'a']
df.loc['a','B']
df.iloc[0,1]
```

### 2) iloc에 슬라이싱 적용

#### 형식:  df.iloc[시작인덱스:끝인덱스:슬라이싱간격]

- 1열의 0행,1행 선택 : 시리즈 형태로 반환

```python
df
df.iloc[[0,1],1]
```

- 1열의 0행과 1행 선택 : 데이터프레임 형태로 반환

```python
df.iloc[[0,1], 1:2]
```

- 0행의 2번째 열부터 끝열까지 반환

```python
df.iloc[0,2:]
df.iloc[0:1,2:]
```

- 0행의 끝에서 두번째 열 이후까지 반환

```python
df.iloc[0,-2:]
df.iloc[0:1, -2:]
```

#### iloc인덱서 정리

- iloc[행위치,열위치] -> 원소값 반환
- iloc[행위치1:행위치2,열위치1:열위치2] -> 원소 반환: df 반환
- iloc[행위치,열위치1:열위치2] -> 원소반환 :시리즈 반환
- iloc[행위치1:행위치2,열위치] ->원소반환 : 시리즈 반환
