# pandas 데이터 파악과 조작

**분석할 데이터를 수집(확보)하면 데이터의 특징을 파악하고 다루기 쉽게 변형하는 작업을 수행해야 한다**

# #2. 데이터 조작(가공)

- 데이터 개수 세기 : count(), value_counts()
- 데이터 정렬 : sort_values(), sort_index()
- 데이터 순위 : rank()
- 데이터 집계 : 합계(sum()), 평균(mean()), 최대(max()), 최소(min())
- 데이터 삭제 : drop(axis=0/1)
- 결측치 처리 : dropna(axis=0/1, subset, inplace)
- 데이터 변경 : 
    - 자료형 변경 : astype()
    - 수치형 데이터를 범주형 데이터로 변경 : 
        - 구간을 지정하여 범주화 : cut(data, bins, labels)
        - 동일한 개수를 갖도록 범주화 : qcut(data, bins_num, labels)
    - 스케일링(scaling) : 정규화(min-max scaling), 표준화(z-score)
    - 시계열데이터 처리
- 행/열에 동일한 함수 적용 : apply(), applymap(), map()
- 열 재구성 : DataFrame.str.split(), Series.str.get()
- 필터링 : isin()

### 열 분리

```python
df = pd.read_csv('data/NFLX.csv')
df.head(3)
df.info()
df.shape
```

#### 시리즈의 문자열 리스트 인덱싱 : Series.str.get(인덱스)

```python
dates = df['Date'].str.split('-')
df['Year'] = dates.str[0]
df['Month'] = dates.str[1]
df['Day'] = dates.str[2]
```

```python
df.head(2)
```

```python
df = df.drop(columns=['Year', 'Month', 'Day'])
```

#### ['year','month','day']로 분리된 시리즈문자열 요소를 데이터프레임 열로 추가

```python
# dates.str.get(0)
# df['Year'] = dates.str.get(0)
# df['Month'] = dates.str.get(1)
# df['Day'] = dates.str.get(2)

# apply를 이용한 분리
new_df = dates.apply(pd.Series)
new_df.columns = ['Year','Month','Day']
new_df.index = df.index
df = df.join(new_df)
df.head(3)
```

```python
df['Date_'] = df['Date'].astype('datetime64[ns]')
```

```python
df = df.drop(columns=['Date_'])
```

```python
df.head(2)
```

```python
df['Date_'] = pd.to_datetime(df['Date'], format='%Y-%m-%d', errors='coerce')
df.head(2)
```

#### 문제. Date컬럼을 날짜형식 데이터로 변환한 후 연, 월, 일 컬럼을 추가

```python
df.info()
```

```python
df['연'] = df['Date_'].dt.year
df['월'] = df['Date_'].dt.month
df['일'] = df['Date_'].dt.day
df['요일'] = df['Date_'].dt.dayofweek
df.head(2)
```

```python
df['요일'] = df['요일'].map({0:'월',1:'화',2:'수',3:'목',4:'금',5:'토',6:'일'})
df.head(3)
```

- .dt 접근자 : 반드시 datetime 타입에서만 사용 가능
- 대용량데이터는 to_datetime() 함수 사용시 'format=' 지정이 유리(astype사용한 경우보다 속도 빠를 수 있음)
- astype()을 이용한 날짜형식 변경 시  속도는 빠르나 에러 처리 옵션이 없음

## 9. 필터링

### 불린 인덱싱(boolean indexing)

```python
titanic.tail()
```

#### 나이가 10대(10~19세) 승객 추출

```python
titanic['age'].fillna(-999)
```

```python
#titanic[(titanic['age'] >= 10) & (titanic['age'] <= 19)]

mask = (titanic['age'] >= 10) & (titanic['age'] <= 19)
mask
#titanic[mask]
titanic.loc[mask,:]
```

#### 나이가 10세 미만이고 여성인 승객 추출

```python
titanic[(titanic['age'] < 10) & (titanic['sex'] == 'female')].head(3)
```

#### 나이가 10세 미만 또는 60세 이상인 승객의 age, sex, alone 열 추출

```python
titanic[(titanic['age'] < 10) | (titanic['age'] >= 60)][['age', 'sex', 'alone']]
```

### isin() 메서드 활용

- isin() 메서드를 적용하면 특정 값을 가진 행들을 추출함
- 형식 : df.isin(추출값 리스트)

#### 함께 탑승한 형제 또는 배우자 수가 3,4,5인 승객만 추출

- 불린 인덱싱 사용

```python
mask = (titanic.sibsp == 3) | (titanic.sibsp == 4) | (titanic.sibsp == 5)
```

```python
titanic[mask]
```

- isin() 메서드 사용

```python
mask = (titanic.sibsp.isin([3,4,5]))
titanic[mask]
```
