# 모집단과 표본분포

#### 모집단과 표본

- 모집단(population) 
    - 조사 대상이 되는 관측 가능한 개체로 된 집단 전체  
- 표본(sample)
    - 모집단에서 선택된 모집단 구성단위의 일부

#### 랜덤표본(random sample)

- 모집단에서 표본으로 뽑힐 가능성을 동일하게 하여 추출된 표본
- 무작위 추출(임의추출) 일부

#### 복원추출과 비복원추출

- 복원추출: 여러 차례 동일한 표본을 선택하는 방법
- 비복원추출: 동일한 표본은 한 번만 선택하는 방법

### 표본분포란?

- 여러 표본들 혹은 재표본들로부터 얻은 표본통계량의 분포
- 예. 표본평균의 분포, 표본분산의 분포 등

- 우리가 다루려는 데이터 대부분은 모집단에서 추출한 표본이며
- 주된 관심은 표본을 통해 모집단을 추측할 수 있는가에 있음
- 만약 표본분포가 어떤 확률분포를 따르는지를 알 수 있다면
- 확률적으로 다룰 수 있으므로 통계적 추론을 통해 모집단의 특성을 추측할 수 있음

#### 복원/비복원 추출함수

```python
np.random.choice(a, size=None, replace=True, p=None)
- a : 배열이면 배열 자체, 정수이면 arange(a)명령으로 배열 생성
- size : 정수, 샘플 숫자
- replace : boolean. True이면 복원추출, False면 비복원추출
- p : 배열, 각 데이터가 선택될 수 있는 확률
```

````python
# 복원추출 : replace=True (디폴트)
for _ in range(4):
    print(np.random.choice([1,2,3,4,5,6], size=6))
````

````python
# 비복원추출 : replace=False
for _ in range(6):
    print(np.random.choice([1,2,3,4,5,6], size=6, replace=False))
    #print(np.random.choice([1,2,3,4,5,6], size=7, replace=False))
    #ValueError: Cannot take a larger sample than population when 'replace=False'
````

````python
np.random.seed(123)
````

#### scores 데이터에서 표본추출

````python
np.random.choice(df['score'], size=10)
````

- 전체 성적 데이터의 평균 : 모평균(μ)

````python
df['score'].mean()
````

- 표본 평균

````python
sample = np.random.choice(df['score'], size=10)
np.mean(sample)
````

````python
s_means = []
for i in range(10):
    sample = np.random.choice(df['score'], size=10)
    s_mean = sample.mean()
    s_means.append(s_mean)
    print(f'{i+1}번째 무작위 추출한 표본의 평균: {s_mean}')
s_means
````

- 표본평균의 평균

````python
np.mean(s_means)
````

- 표본평균의 표준편차

````python
np.std(s_means)
````

- 표본추출함수 작성

````python
def sampling(data, n, loop):
    s_means = []
    for i in range(loop):
        sample = np.random.choice(data, n)
        s_mean = sample.mean()
        s_means.append(s_mean)    
    return s_means

def sample_mean_std(data, n, loop):
    s_means = sampling(data, n, loop)
    print(f'n={n}인 표본{loop}개의 표본평균의 평균:{np.mean(s_means):.2f},\
    표준편차:{np.std(s_means):.2f}')
````

````python
sample_mean_std(df['score'], n=10, loop=100)
sample_mean_std(df['score'], n=10, loop=1000)
sample_mean_std(df['score'], n=10, loop=10000)
````

````python
sample_mean_std(df['score'], n=20, loop=100)
sample_mean_std(df['score'], n=20, loop=1000)
sample_mean_std(df['score'], n=20, loop=10000)
sample_mean_std(df['score'], n=50, loop=100)
sample_mean_std(df['score'], n=50, loop=1000)
sample_mean_std(df['score'], n=50, loop=10000)
````

#### 가능한 모든 표본으로 이루어진 표본평균들의 평균은 모집단의 평균과 같다

$ E(\bar X) = \mu $

- 복원추출로 이루어진 가능한 모든 표본들의 표본평균

````python
population = [1,2,3]
n=2
# 복원추출로 이루어진 가능한 모든 표본들 -> 9가지
samples=np.array([[1,1],[1,2],[1,3],
                  [2,1],[2,2],[2,3],
                  [3,1],[3,2],[3,3]])
sample_means = samples.mean(axis=1)
sample_means
````

````python
# 표본평균들의 평균
sample_means.mean()
````

````python
# 모평균
np.mean(population)
````

=> 모평균 = 표본평균들의 평균

````python
# 표본평균들의 표본편차
np.std(sample_means), sample_means.std()
````

````python
# 모평균의 표준편차
np.std(population)
````

````python
# 표준오차(Standart Error: S.E.) : 모표준편차 / sqrt(표본수)
np.std(population) / np.sqrt(n)
````

[정리] 복원추출인 경우

- 가능한 모든 표본의 표본평균들에 대한 평균은 모평균과 같음
- 표본평균들의 표준편차는 모집단의 표준편차를 샘플수의 제곱근으로 나눈 것과 같음

**비복원추출로 이루어진 가능한 모든 표본들의 표본평균**

````python
population = [1,2,3]
n=2  # 표본수

# 비복원추출로 이루어진 가능한 모든 표본들 -> 6가지
samples=np.array([[1,2],[1,3],
                  [2,1],[2,3],
                  [3,1],[3,2]])
sample_means = samples.mean(axis=1)
sample_means
````

````python
# 표본평균들의 표본편차
np.mean(population), np.mean(sample_means)
````

=> 비복원 추출로 추출한 경우에도 모평균과 같아짐

#### 표본자료의 분포

````python
sample1 = np.random.choice(df['score'], 100)
sample2 = np.random.choice(df['score'], 1000)
sample3 = np.random.choice(df['score'], 10000)
fig = plt.figure(figsize=(14,10))

# 모집단 분포(전체 성적 데이터)
ax = fig.add_subplot(221)
ax.hist(df['score'], bins=100, range=(0,100), density=True)
ax.set_title('Population Dist.')
ax.set_xlim(20,100)
ax.set_xlabel('score')
ax.set_ylabel('relative freq.')

# 표본 데이터의 분포(n=100)
ax = fig.add_subplot(222)
ax.hist(sample1, bins=100, range=(0,100), density=True)
ax.set_title('Sample Dist.(n=100)')
ax.set_xlim(20,100)
ax.set_xlabel('score')
ax.set_ylabel('relative freq.')

# 표본 데이터의 분포(n=1000)
ax = fig.add_subplot(223)
ax.hist(sample2, bins=100, range=(0,100), density=True)
ax.set_title('Sample Dist.(n=1000)')
ax.set_xlim(20,100)
ax.set_xlabel('score')
ax.set_ylabel('relative freq.')

# 표본 데이터의 분포(n=10000)

ax = fig.add_subplot(224)
ax.hist(sample3, bins=100, range=(0,100), density=True)
ax.set_title('Sample Dist.(n=10000)')
ax.set_xlim(20,100)
ax.set_xlabel('score')
ax.set_ylabel('relative freq.')

plt.tight_layout()
plt.show()
````

#### 표본평균의 분포

````python
sample_means = sampling(df['score'], 20, 10000)
fig = plt.figure(figsize=(6, 4))
ax = fig.add_subplot(111)
ax.hist(sample_means, bins=100, range=(0, 100), density=True)

# 모평균을 세로선으로 표시
ax.vlines(np.mean(df['score']), 0, 1, 'gray')
ax.set_xlim(50, 90)
ax.set_ylim(0, 0.13)
ax.set_xlabel('mean of score')
ax.set_ylabel('relative frequency')
plt.show()
````
