# 모집단과 표본분포

#### 모집단과 표본

- 모집단(population) 
    - 조사 대상이 되는 관측 가능한 개체로 된 집단 전체  
- 표본(sample)
    - 모집단에서 선택된 모집단 구성단위의 일부

#### 랜덤표본(random sample)

- 모집단에서 표본으로 뽑힐 가능성을 동일하게 하여 추출된 표본
- 무작위 추출(임의추출) 일부

#### 복원추출과 비복원추출

- 복원추출: 여러 차례 동일한 표본을 선택하는 방법
- 비복원추출: 동일한 표본은 한 번만 선택하는 방법

### 표본분포란?

- 여러 표본들 혹은 재표본들로부터 얻은 표본통계량의 분포
- 예. 표본평균의 분포, 표본분산의 분포 등

- 우리가 다루려는 데이터 대부분은 모집단에서 추출한 표본이며
- 주된 관심은 표본을 통해 모집단을 추측할 수 있는가에 있음
- 만약 표본분포가 어떤 확률분포를 따르는지를 알 수 있다면
- 확률적으로 다룰 수 있으므로 통계적 추론을 통해 모집단의 특성을 추측할 수 있음

## 표본평균의 분포

### 예제. 전교생 수학 성적

#### 복원/비복원 추출함수

```python
np.random.choice(a, size=None, replace=True, p=None)
- a : 배열이면 배열 자체, 정수이면 arange(a)명령으로 배열 생성
- size : 정수, 샘플 숫자
- replace : boolean. True이면 복원추출, False면 비복원추출
- p : 배열, 각 데이터가 선택될 수 있는 확률
```

#### scores 데이터에서 표본추출

- 전체 성적 데이터의 평균 : 모평균(μ)

- 표본 평균

- 표본평균의 평균

- 표본평균의 표준편차

- 표본추출함수 작성

#### 가능한 모든 표본으로 이루어진 표본평균들의 평균은 모집단의 평균과 같다
$ E(\bar X) = \mu $


**복원추출로 이루어진 가능한 모든 표본들의 평균분포**

=> 가능한 모든 표본들에 대한 평균들의 평균은 모평균과 같아지며,

=> 표본평균들의 표준편차는 모집단의 표준편차를 샘플수의 제곱근으로 나눈 것과 같다

**비복원추출로 이루어진 가능한 모든 표본들의 평균분포**

=> 비복원추출로 뽑힌 표본들의 평균의 평균도 모평균과 같아진다.
