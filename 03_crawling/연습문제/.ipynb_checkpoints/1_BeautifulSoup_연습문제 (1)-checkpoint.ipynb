#### 문제 4. 도서 상세정보 추출

````python
book_title = soup.select('.book h2.title')
title = [ t.text for t in book_title ]
book_author = soup.select('.book p.author')
author = [ a.text for a in book_author ]
book_publisher = soup.select('.book p.publisher')
publisher = [ p.text for p in book_publisher ]
book_price = soup.select('.book span.price')
price = [ p.text for p in book_price ]
book_rating = soup.select('.book span.rating')
rating = [ r.text for r in book_rating ]

book = {
    '제목' : title,
    '저자' : author,
    '출판사' : publisher,
    '가격' : price,
    '평점' : rating
}
book
````

#### 문제 8. 가격 데이터 정제 및 조건 검색

가격은 현재 `"22,000원"`과 같은 문자열입니다.

1. `,`와 `원`을 제거합니다.
2. `int()`를 이용하여 정수로 변환합니다.
3. 가격이 **30,000원 이상인 도서**만 출력합니다.

**실행 결과**

```text
파이썬 데이터 분석 : 30000원
머신러닝 입문 : 35000원
딥러닝과 생성형 AI : 42000원
```

````python
book = soup.select('.book')
for b in book:
    book_title = b.select_one('.book h2.title').text
    price_to_num = b.select_one('.price').text[:-1].replace(",","")
    price = int(price_to_num)
    # print(f'{book_title} : {price}원')
    if price >= 30000:
        print(f'{book_title} : {price}원')
        
    

        # <div class="book" data-category="ai">
        #     <h2 class="title"><a href="/book/105">딥러닝과 생성형 AI</a></h2>
        #     <p class="author">장영실</p>
        #     <p class="publisher">AI북스</p>
        #     <span class="price">42,000원</span>
        #     <span class="rating">4.9</span>
````

#### 문제 9. 평점 조건 검색

평점을 `float`형으로 변환하고 **4.8 이상인 도서**만 출력하세요.

**실행 결과**

```text
혼자 공부하는 파이썬 : 4.8
파이썬 데이터 분석 : 4.9
딥러닝과 생성형 AI : 4.9
```

````python
book = soup.select('.book')
for b in book:
    book_title = b.select_one('.book h2.title').text
    book_rating = float(b.select_one('.rating').text)
    
    if book_rating >= 4.8:
        print(f'{book_title} : {book_rating}')
        
        # <div class="book" data-category="ai">
        #     <h2 class="title"><a href="/book/105">딥러닝과 생성형 AI</a></h2>
        #     <p class="author">장영실</p>
        #     <p class="publisher">AI북스</p>
        #     <span class="price">42,000원</span>
        #     <span class="rating">4.9</span>
````

#### 문제 10. 크롤링 데이터를 리스트와 딕셔너리로 저장

각 도서의 정보를 다음 구조로 `book_list`에 저장하세요.

```python
[
    {
        "title": "혼자 공부하는 파이썬",
        "author": "홍길동",
        "publisher": "한빛출판사",
        "category": "python",
        "price": 22000,
        "rating": 4.8,
        "link": "/book/101"
    },
    ...
]
```

가격은 `int`, 평점은 `float`로 저장합니다.

````python
book_list = []
for b in soup.select('.book'):
    title = b.select_one('.book h2.title').text
    
    author = b.select_one('.book p.author').text
    publisher = b.select_one('.book p.publisher').text
    catagory = b['data-category']
    
    book_price = b.select_one('.book span.price')
    price = int(book_price.text[:-1].replace(",",""))
    rating = float(b.select_one('.book span.rating').text)
    link = b.select_one('a')['href']
    
    book_info = {
        'title' : title,
        'author' : author,
        'publisher' : publisher,
        "category": catagory,
        'price' : price,
        'rating' : rating,
        "link": link
    }
    book_list.append(book_info)


book_list
        # <div class="book" data-category="ai">
        #     <h2 class="title"><a href="/book/105">딥러닝과 생성형 AI</a></h2>
        #     <p class="author">장영실</p>
        #     <p class="publisher">AI북스</p>
        #     <span class="price">42,000원</span>
        #     <span class="rating">4.9</span>
````

#### 문제 11. 수집 데이터 분석

앞에서 생성한 `book_list`를 이용하여 다음 항목을 구하세요.

1. 전체 도서 수
2. 전체 도서의 평균 가격
3. 가장 비싼 도서
4. 최고 평점
5. 최고 평점을 받은 모든 도서
6. 카테고리별 도서 수

**기대 결과 일부**

```text
전체 도서 수 : 5권
평균 가격 : 31,400원
가장 비싼 도서 : 딥러닝과 생성형 AI / 42,000원
최고 평점 : 4.9
```

````python
book_num = len(book_list)

book_price = []
# book_price = [ b['price'] for b in book_list ]

for b in book_list:
    book_price.append(b['price'])

book_price_avg = sum(book_price)/len(book_price)

most_exp_book = sorted(book_list, key = lambda x:x['price'])[-1]
highest_rate = max(book_list, key = lambda x:x['rating'])['rating']

print(f'전체 도서 수 : {book_num}')
print(f'평균 가격 : {book_price_avg}')
print(f'가장 비싼 도서 : {most_exp_book['title']} / {most_exp_book['price']}')
print(f'최고 평점 : {highest_rate}')

    
````

### 최종 프로그램 요구사항

- 전체 도서 목록을 번호와 함께 출력
- 가격은 천 단위 쉼표를 포함하여 출력
- 전체 도서 수 출력
- 평균 가격 출력
- 최고 가격 도서 출력
- 최고 평점과 해당 도서 출력
- `python`, `database`, `ai` 카테고리별 도서 수 출력

````python
# HTML
html = '''
<!DOCTYPE html>
<html lang="ko">
<head>
    <meta charset="UTF-8">
    <title>AI 온라인 서점</title>
</head>
<body>
<div id="container">
    <h1>AI 온라인 서점</h1>

    <div class="book-list">

        <div class="book" data-category="python">
            <h2 class="title"><a href="/book/101">혼자 공부하는 파이썬</a></h2>
            <p class="author">홍길동</p>
            <p class="publisher">한빛출판사</p>
            <span class="price">22,000원</span>
            <span class="rating">4.8</span>
        </div>

        <div class="book" data-category="database">
            <h2 class="title"><a href="/book/102">MySQL 데이터베이스</a></h2>
            <p class="author">이순신</p>
            <p class="publisher">정보출판사</p>
            <span class="price">28,000원</span>
            <span class="rating">4.5</span>
        </div>

        <div class="book" data-category="python">
            <h2 class="title"><a href="/book/103">파이썬 데이터 분석</a></h2>
            <p class="author">김유신</p>
            <p class="publisher">한빛출판사</p>
            <span class="price">30,000원</span>
            <span class="rating">4.9</span>
        </div>

        <div class="book" data-category="ai">
            <h2 class="title"><a href="/book/104">머신러닝 입문</a></h2>
            <p class="author">강감찬</p>
            <p class="publisher">AI북스</p>
            <span class="price">35,000원</span>
            <span class="rating">4.7</span>
        </div>

        <div class="book" data-category="ai">
            <h2 class="title"><a href="/book/105">딥러닝과 생성형 AI</a></h2>
            <p class="author">장영실</p>
            <p class="publisher">AI북스</p>
            <span class="price">42,000원</span>
            <span class="rating">4.9</span>
        </div>

    </div>
</div>
</body>
</html>
'''

soup = BeautifulSoup(html, "html.parser")
print("실습용 HTML 로드 완료")
````

````python
# list 저장
book_list = []
for b in soup.select('.book'):
    title = b.select_one('.book h2.title').text
    
    author = b.select_one('.book p.author').text
    publisher = b.select_one('.book p.publisher').text
    catagory = b['data-category']
    
    book_price = b.select_one('.book span.price')
    price = int(book_price.text[:-1].replace(",",""))
    rating = float(b.select_one('.book span.rating').text)
    link = b.select_one('a')['href']
    
    book_info = {
        'title' : title,
        'author' : author,
        'publisher' : publisher,
        "category": catagory,
        'price' : price,
        'rating' : rating,
        "link": link
    }
    book_list.append(book_info)
````

````python
# 데이터 프레임 저장
import pandas as pd
book_df = pd.DataFrame(book_list)
book_df
````

````python
#전체 도서 목록을 번호와 함께 출력
book_df
````

````python
#가격은 천 단위 쉼표를 포함하여 출력
book_df['price'] = book_df['price'].apply(lambda p: f"{p:,}원")
book_df
````

````python
#전체 도서 수 출력
len(book_list)
````

````python
# 평균 가격 출력
sum( [ b['price'] for b in book_list ] ) / len(book_list)
````

````python
#최고 가격 도서 출력
# def key1(x):
#     return x['price']
# exp_book = sorted(book_list, key = key1)[-1]
# #exp_book = sorted(book_list, key = lambda x: x['price'])[-1]
# exp_book
exp_price = max( b['price'] for b in book_list)
exp_book = [ b for b in book_list if b['price'] == exp_price ]
exp_book = pd.DataFrame(exp_book)

book_df[ book_df['price'] == max( book_df['price'])]
````

````python
#최고 평점과 해당 도서 출력
highest_rating = max( b['rating'] for b in book_list)
highest_book = [ b for b in book_list if b['rating'] == highest_rating ]
highest_book = pd.DataFrame(highest_book)


book_df[ book_df['rating'] == max( book_df['rating'])]
````

````python
# `python`, `database`, `ai` 카테고리별 도서 수 출력
cat = 'python'
print( len( [ b for b in book_list if b['category'] == cat ] ) )

cat = 'database'
print( len( [ b for b in book_list if b['category'] == cat ] ) )

cat = 'ai'
print( len( [ b for b in book_list if b['category'] == cat ] ) )
````
