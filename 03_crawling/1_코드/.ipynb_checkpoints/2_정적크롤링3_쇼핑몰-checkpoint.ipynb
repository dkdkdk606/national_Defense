#### 크롤링 페이지 읽어오기

```python
url = 'https://jolse.com/category/toners-mists/1019/?page=2'
html = request.urlopen(url).read()
soup = BeautifulSoup(html, 'html.parser')
```

#### 1번 페이지의 전체 상품의 상품명, 정상가, 세일가 추출

```python
# 이름 <span class="" style="font-size:14px;color:#555555;font-weight:bold;">numbuzin No.1 Pure-Full Calming Herb Toner 300ml</span>
# 정상가 <span class="" style="font-size:16px;color:#888888;text-decoration:line-through;">USD 28.00</span>
# 세일가 <span class="" style="font-size:20px;color:#ff2d46;font-weight:bold;">USD 19.60 <span style="font-size:12px;color:#4d4d4d;">(30% OFF)</span></span>
# 상품링크 
'''
<div class="prdImg">
			<a href="/product/numbuzin-no1-pure-full-calming-herb-toner-300ml/61576/category/1019/display/1/" name="anchorBoxName_61576" class=" _evt_tracker0"><img src="https://cafe24img.poxo.com/jolsejolse/web/product/medium/202509/c1d5c0ddff63a5c4a30ece77ab14b002.jpg" id="eListPrdImage61576_1" alt="numbuzin No.1 Pure-Full Calming Herb Toner 300ml bottle with a clear watery texture"></a>
			<span class="wish"><img src="/web/upload/icon_202112211850391800.png" class="icon_img ec-product-listwishicon" alt="Before add to wish list" productno="61576" categoryno="1019" icon_status="off" login_status="F" individual-set="F"></span>
			<span class="cart"><img src="/web/upload/icon_202112211850286500.png" onclick="category_add_basket('61576','1019', '1', 'A0000', false, '1', 'P000DNCI', 'B', 'F', '0');" alt="Add to cart" class="ec-admin-icon cart"></span>
		</div>
        '''
# 메인 박스  <ul class="prdList grid5">
```

```python

main_box = soup.select_one('ul.prdList.grid5')
# item_name = main_box.select('span[style="font-size:14px;color:#555555;font-weight:bold;"]').text.strip()
item_names = main_box.select('span[style="font-size:14px;color:#555555;font-weight:bold;"]')
item_n_prices = main_box.select('span[style="font-size:16px;color:#888888;text-decoration:line-through;"]')
item_s_prices = main_box.select('span[style="font-size:20px;color:#ff2d46;font-weight:bold;"]')
item_links = main_box.select('div.prdImg a')

item_name = [ item.text.strip() for item in item_names if item.text.strip() != 'Product Name' ]
item_n_price = [ item.text.strip() for item in item_n_prices]
item_s_price = [ item.text.strip() for item in item_s_prices if item.text.strip() !=""]
item_link = [ 'https://jolse.com' + a['href'] for a in item_links ]
```

####  데이터프레임으로 저장하고 아래와 같이 출력

```python

item_df = pd.DataFrame(data = {'이름': item_name, '정상가':item_n_price, '세일가':item_s_price, '상품링크':item_link})
item_df.to_csv('jolse_category_tonersmists.csv', index=False)

item_df
```

```python
import random
```

```python
def get_url_list(url):
    url = 'https://jolse.com/category/toners-mists/1019/?page=2'
    html = request.urlopen(url).read()
    soup = BeautifulSoup(html, 'html.parser')

    main_box = soup.select_one('ul.prdList.grid5')

    item_links = main_box.select('div.prdImg a')
    item_link = [ 'https://jolse.com' + a['href'] for a in item_links ]

    return item_link

def get_item(url):
    html = request.urlopen(url).read()
    soup = BeautifulSoup(html, 'html.parser')

    item_name = soup.select_one('.headingArea h2').text.strip()
    item_n_price = soup.select_one('strong#span_product_price_text').text.strip()
    try:
        item_s_price = soup.select_one('span#span_product_price_sale').text.strip()
    except:
        item_s_price = '세일중이지 않음'

    item_dict = {'name':item_name,
                 'price':item_n_price,
                 'sale_price':(item_s_price if item_s_price else "세일중이지 않음" ),
                 'link': url
                }
        
    return item_dict
    
print(print_item(get_url_list('https://jolse.com/category/toners-mists/1019/?page=2')[random.randint(0, 39)]))


    # item_name = main_box.select_one('span[style="font-size:14px;color:#555555;font-weight:bold;"]').text.strip()
    # item_n_price = main_box.select_one('span[style="font-size:16px;color:#888888;text-decoration:line-through;"]').text.strip()
    
    # item_link = 'https://jolse.com' + main_box.select_one('div.prdImg a').text.strip()
        
    # is_sale = main_box.select_one('span[style="font-size:20px;color:#ff2d46;font-weight:bold;"]')
    # item_s_price = is_sale.text.strip() if is_sale else ''
```

#### 2. 1번에서 작성한 함수를 사용하여 한 페이지에 들어있는 상품목록 추출하는 함수 작성하시오.

```python
def get_item_list(url):

    item_list = []
    url_list = get_url_list(url)
    #print(url_list)
    
    #item_list = [ u for u in get_item(url_list) ]
    for u in url_list:
        item_list.append(get_item(u))
        
    return item_list


#print(get_item_list('https://jolse.com/category/toners-mists/1019/'))
```

#### 3. 지정한 페이지의 전체 상품목록(상품명, 정상가, 세일가, 상품링크)을 데이터프레임으로 생성하는 함수를 작성하시오.

```python
def get_item_df(url):
    item_list = get_item_list(url)
    item_df = pd.DataFrame(item_list, index = range(1, len(item_list)+1))
    return item_df
```
