## 정적크롤링 실습1. 명언 수집

- 명언을 제공하는 웹사이트에서 'life'관련 명언을 스크랩하여 엑셀 파일에 저장하기
- 대상 : http://quotes.toscrape.com/

#### 크롤링하는 웹사이트 살펴보기

- 사이트에서 life 태그 클릭한 후 명언의 (about) 클릭

- 명언들 페이지의 하단 부분의 [Next]버튼, [Previeous]버튼

- 살펴본 명언 웹사이트의 특징
    - 명언들이 태그별로 모여 있음
    - 한 페이지에 일정 수의 명언만 보이고 나머지는 다음(Next) 페이지를 통해 볼 수 있음

#### 웹사이트 구조 확인 : 개발자도구(F12)

#### 데이터 수집을 위한 절차

- 1단계 : 페이지 구조 확인 및 명언/저자가 있는 위치(태그, CSS선택자 등) 파악
- 2단계 : 웹페이지 불러오기(읽기)(url 요청 -> 응답)
- 3단계 : 각 태그(life)에 맞는 명언/저자 추출 (BeautifulSoup 사용)

**웹페이지 불러오기**

```python
# 모듈 임포트
import requests
from urllib.request import urlopen
from bs4 import BeautifulSoup
```

```python
# life 명언 페이지 HTML 받아오기
url = 'https://quotes.toscrape.com/tag/life/'
html = urlopen(url).read()
soup = BeautifulSoup(html, 'html.parser')
```

**각 태그에 맞는 명언/저자 추출(파싱)**

```python
# 명언과 저자가 있는 부분 선택하여 추출 : <div class='qoute'> => div.qoute
quotes = soup.find_all('div', {'class' : 'quote'})
print(f'명언개수: {len(quotes)}')
```

```python
quotes[0]
```

```python
quotes[0].find('span', {'class':'text'}).text
```

```python
quotes[0].find('span', {'class':'text'}).get_text()
```

```python
quotes[0].find('span', {'class':'text'}).get_text().strip('“').strip('”')
```

```python
quotes[0].find('small', {'class':'author'}).text
```

```python
texts, authors = [], []
for quote in quotes:
    texts.append(quote.find('span', {'class':'text'}).get_text().strip('“').strip('”'))
    authors.append(quote.find('small', {'class':'author'}).get_text(strip=True))
```

```python
authors
```

```python
texts1, authors1 = [], []
texts1 = [quote.find('span', {'class':'text'}).get_text().strip('“').strip('”') for quote in quotes]
authors1 = [quote.find('small', {'class':'author'}).get_text(strip=True)  for quote in quotes ]
print(f'저자: {authors1},\n\n 명언{texts1}')
```

```python
# 딕셔너리 리스트로 저장
quote_dict = { 'texts':[], 'authors':[] }
for quote in quotes:
    quote_dict['texts'].append(quote.find('span', {'class':'text'}).get_text().strip('“').strip('”'))
    quote_dict['authors'].append(quote.find('small', {'class':'author'}).get_text(strip=True))
```

```python
quote_dict
```

```python
#저자, 명언 튜플로 저장
quote_list = []
for quote in quotes:
  text = quote.find('span', {'class': 'text'}).get_text().strip('“').strip('”')
  author = quote.find('small', {'class': 'author'}).get_text(strip=True)
  quote_list.append((author, text))
quote_list
```
