# selenium을 이용한 동적크롤링 실습

### 웹페이지 크롤링 프로세스

1. 크롤링하려는 웹사이트 접속
2. 분석하려는 페이지를 찾고, 페이지 분석을 통해 어떤 태그와 속성이 사용되었는지 파악
3. 파이썬을 이용해 필요 데이터 추출
4. 추출한 데이터를 기반으로 분석 진행

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

## #1. 네이버 로그인

### 브라우저에 맞는 웹드라이버 객체 생성

```python
url = 'https://nid.naver.com/nidlogin.login'
driver = webdriver.Chrome()
driver.get(url)
```

```python
ipswitch = driver.find_element(By.CSS_SELECTOR, 'input#switchIP')
ipswitch_btn = driver.find_element(By.CSS_SELECTOR, 'label.slider')
is_ipswitch = ipswitch.get_attribute(
    'aria-checked'
)
print(is_ipswitch)

if is_ipswitch == 'true':
    ipswitch_btn.click()

```

### 네이버 로그인 페이지 접속

```python
id_box = driver.find_element(By.CSS_SELECTOR, 'input#id')
# id_label_box = driver.find_element(By.CSS_SELECTOR, 'label[for="id"]')
time.sleep(0.3)





```

### 동적 로그인 진행

#### 로그인 아이디 정보 전달

```python
#<input id="id" name="id" value="" type="text" aria-label="아이디 또는 전화번호" class="input_text" data-i18n="inputID" maxlength="41" data-input-filter="ascii" placeholder="">
# id_box.click()
id_box.send_keys('dkdkdk606')
time.sleep(0.3)
```

#### 로그인 비밀번호 전달

```python
#<input id="pw" name="pw" type="password" aria-label="비밀번호" class="input_text" data-i18n="inputPW" maxlength="16" data-input-filter="ascii">
pw_box = driver.find_element(By.CSS_SELECTOR, 'input#pw')
time.sleep(0.3)
# pw_box.click()
pw_box.send_keys('####################')
```

#### XPath를이용하여 버튼 클릭 위치 지정하고 로그인 버튼 클릭

##### XPath
- XML Path Language
- https://ko.wikipedia.org/wiki/XPath
- W3C의 표준으로 확장 생성 언어 문서의 구조를 통해 경로 위에 지정한 구문을 사용하여 항목을 배치하고 처리하는 방법을 기술하는 언어
- XML 문서의 노드를 정의하기 위하여 경로식을 사용
- 웹페이지 내용을 XML 형식으로 보여줌
- 접근 가능한 위치 정보를 확인할 수 있음

### 자동입력 방지 문자 우회하여 로그인 진행 방법

- 자동입력 방지 문자 우회하기 위해
- 브라우저 종료 후 다시 접속
- send_keys() 함수 대신에 직접 id/pass값을 대입하는 script 코드를 작성 후 webdriver 통해 실행

#### webdriver 통해 script 코드 직접 실행하여 로그인
- **driver.execute_script()** 함수 - driver로 script 코드를 직접 실행

```python
time.sleep(0.3)

#<button type="button" class="btn_done" id="loginBtn_row"> <span class="text" data-i18n="btnLogin">로그인</span></button>
login_box = driver.find_element(By.CSS_SELECTOR, 'button#loginBtn_row')
login_box.click()
```
