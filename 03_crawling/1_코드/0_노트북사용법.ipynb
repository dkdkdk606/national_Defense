# Jupyter notebook 사용법

### 노트북 요소
- 셀(cell) : 코드, 마크다운, raw 등의 내용을 담음

### 노트북 모드
#### 편집모드 : 코드나 문서를 작성하기 위한 모드
- 표시 : 셀 경계선이 파란색으로 변하고, 편집영역은 흰식, 프롬프트 깜빡임
- 진입방법 : 명령모드에서 Enter키를 누르거나 셀의 편집영역을 마우스로 클릭

#### 명령모드 : 셀을 다루기 위한 모드
- 표시 : 셀 경계산와 셀 영역의 색이 회색이 됨
- 진입방법 : 편집모드에서 Esc키를 누르거나 셀의 편집영역의 바깥 부분을 클릭

### 노트북 단축키
**셀 실행**
- Ctrl + Enter  : 현재 셀을 수행, 위치는 현재 셀이 머물러 있음
- Shift + Enter : 현재 셀을 수행하고 아래셀을 선택, 아래에 셀이 없으면 새로운 셀 추가
- Alt + Enter   : 현재 셀을 수행하고 아래에 새로운 셀을 추가

**셀 이동**
- 위, 아래 방향키 :
    - 편집모드일 때 : 현재 셀에서 현재 커서가 있는 줄의 위/아래 줄로 이동
    - 명령모드일 때 : 위/아래 셀로 이동
- 왼쪽, 오른쪽 방향키 :
    - 편집모드일 때 :  현재 셀에서 현재 커서가 있는 왼쪽이나 오른쪽 칸으로 이동
    - 명령모드일 때 : 마크다운의 제목, 본문을 축소/펼치기 형식으로 보여줌

**노트북 저장**
- Ctrl+S : 현재 이름으로 저장
- Ctrl+Shift+S : 다른 이름으로 저장

**명령(선택) 모드**

: 셀의 색이 회색이 된 경우
- 코드 셀에서 Esc 또는 Ctrl+m을 누르면 명령모드 전환
- a : 현재 셀 위(above)에 새로운 셀 추가
- b : 현재 셀 아래(below)에 새로운 셀 추가
- dd : 현재(선택) 셀 삭제
- c : 현재(선택) 셀 복사
- v : 현재(선택) 셀 붙여넣기
- x : 현재(선택) 셀 잘라내기
- o : 실행 결과 열기/닫기
- y : 코드 셀로 변경
- m : 마크다운 셀로 변경
- Enter : 현재셀의 입력모드로 돌아가기
- 상/하 방향키 : 셀의 위/아래 셀로 이동

**편집 모드**

: 커서가 있는 셀 테두리가 파란색이 된 상태
- Shift+Ctrl+[-] : 커서 위치에서 셀을 둘로 나누기(분할)
- Ctrl+a : 셀의 전체 선택
- Ctrl+z : 실행 취소
- Ctrl+y : 다시 실행
- Ctrl+/ : 주석/비주석 처리
- Esc : 명령모드로 전환
- Ctrl+Enter, Shift+Enter, Alt+Enter : 셀 실행

### 코드 셀 작성과 실행

- 코드 셀 실행
- 주석 설정
- 마크다운 셀로 변경
- 코드 셀에 작성된 코드들은 위에서 아래 순서로 실행
- 앞의 셀에서 실행된 결과(변수, 함수, 객체 등)는 뒤의 셀에서 사용 가능
- *주의* : 노트북 **커널**이 재시작되면 앞의 셀의 결과는 사용할 수 없음. 처음부터 다시 실행해야 함
- 커널 재시작 / 중단 / 커널재시작+모든결과삭제 / 커널재시작+현재셀까지실행 / 커널재시작+모든셀실행
- 함수 도움말 보여주기 : 코드셀에서 함수이름 뒤에 커서를 두고 Shift+Tap 키를 누름

### 마크다운 작성
- 제목 마크다운 :  #, ##, ###, ####, #### + 제목글 (heading)
- 강조(굵게:bold) : ** 내용 **
- 기울임 : * 내용  *
---
- 가로줄 : ---  hr 태그와 같음
- 목록 표시 : *, -, 숫자를 사용하면 목록으로 표시, 하위 목록은 tab을 주어 표시 가능

- html의 태그를 직접 넣어서 작성 가능
- 수식 기호 삽입이 가능
- 표 삽입
