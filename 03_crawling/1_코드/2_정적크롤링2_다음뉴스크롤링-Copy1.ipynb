### 2. 뉴스 기사 하이퍼링크 주소 추출하기

- 뉴스기사 페이지로 이동하는 \<a\>태그의 하이퍼링크 주소 가져오기

#### 문제. 다음 뉴스 머리기사의 언론사, 제목, 링크를 크롤링하고 그 결과를 데이터프레임으로 생성하기

```python
# 모듈 임포트
from urllib import request
from bs4 import BeautifulSoup
import pandas as pd

# 단계1. 페이지 읽기
url = 'https://news.daum.net'
html = request.urlopen(url).read()

# 단계2. 파싱
soup = BeautifulSoup(html, 'html.parser')

# 뉴스 카드 전체 영역(또는 개별 뉴스 링크 태그)을 기준으로 반복문 돌리기
# 제공해주신 HTML에 따르면 각 뉴스는 a.item_newsheadline2 로 감싸져 있습니다.
news_items = soup.select('a.item_newsheadline2')

title_list = []
media_list = []
link_list = []

for item in news_items:
    # 각 뉴스 박스 안에서 제목 추출
    title_elem = item.select_one('strong.tit_txt')
    # 각 뉴스 박스 안에서 언론사 추출
    media_elem = item.select_one('span.txt_info')
    
    # 요소가 존재하는 경우에만 리스트에 추가
    if title_elem and media_elem:
        title_list.append(title_elem.text.strip())
        media_list.append(media_elem.text.strip())
        link_list.append(item.get('href', ''))

# 수집된 데이터 개수 확인
print(f"수집된 뉴스 건수: {len(title_list)}")

# 단계3. 데이터프레임으로 변환 & 파일 저장
news_df = pd.DataFrame({
    '언론사': media_list, 
    '뉴스제목': title_list, 
    '기사링크': link_list
})

print(news_df)

# 파일 저장 (폴더가 없다면 에러가 날 수 있으니 참고하세요)
# news_df.to_csv('daum_headline_news.csv', index=False, encoding='utf-8-sig')


'''
# 단계1. 페이지 읽기
url = 'https://news.daum.net'
html = request.urlopen(url).read()

# 단계2. 파싱
soup = BeautifulSoup(html, 'html.parser')

main_news = soup.select_one('ul.list_newaheadline2')

titles = main_news.select('strong.tit_txt')
medias = main_news.select('span.txt_info')
links = main_news.select('a.item_newaheadline2')
medias = main_news.select('span.txt_info')            # 언론사
links = main_news.select('a.item_newaheadline2')      # 뉴스 상세페이지링크

title_list = [t.text for t in titles]
media_list = [m.text for m in medias[::2]]
link_list = [a['href'] for a in links]

# 단계3. 데이터프레임으로 변환 & 파일 저장
news_df = pd.DataFrame(data = {'언론사': media_list, '뉴스제목':title_list, '기사링크':link_list})
news_df.to_csv('crawl_data/daum.headline_news.csv', index=False)
news_df


'''

```

### 4. 하이퍼링크된 모든 기사의 제목과 본문 추출하기

```python
def print_news_page(url):
    title, author, date, article = extract_news_page(url)
    print('='*30)
    print(f'기사제목:{title}\n기자:{author}\n작성일:{date}\n\n뉴스내용:{article}')
```

### 추출한 기사내용과 정보를 데이터프레임과 csv파일로 저장

```python
def save_news_data(url_list):
    title_list, author_list, date_list, article_list = [], [], [], []
    for url in url_list:
        title, author, date, article = extract_news_page(url)
        title_list.append(title)
        author_list.append(author)
        date_list.append(date)
        article_list.append(article)
    return {'제목':title_list, '작성자':author_list,
            '작성일':date_list, '뉴스내용':article_list}
```

```python
news_dict = save_news_data(link_list).copy()
news_df = pd.DataFrame(data=news_dict)
news_df.to_csv('crawl_data/daum_news_pages.csv', index=False)
news_df
```
